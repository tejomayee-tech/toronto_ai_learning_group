### Explanation:
- **`merge()`**: This is used to combine two DataFrames based on a common column. In this case, we’re merging on the `born_country` column from the `athlete_df` DataFrame and the `full_country_name` column from the `country_mapping_df` DataFrame.
- **`left_on`** and **`right_on`**: These parameters specify the columns to merge on in the left and right DataFrames, respectively. You’re matching the full country name from the athlete data (`born_country`) with the country name mapping.
- **`how='left'`**: This means that all rows from the left DataFrame (`athlete_df`) will be kept, even if there is no match in the right DataFrame (`country_mapping_df`).

## Filtering Data and Selecting Columns

````python
biosData.loc[biosData["height_cm"]>175] # loc expects a boolean condition (True/False values) or the index of the rows you want to select.
                                        # example of this condition on height provides bolean based on which row is selected or skipped
                                        # unpack the logic by using an bolean list of size equal to the rows count in the bioData  
boolean_array = [True, False, True, False, True]
biosData.loc[boolean_array]
display( biosData.loc[boolean_array][['name','born_date']] ) #first approach
display( biosData.loc[boolean_array,['name','born_date']])   # corrected approach when not using conditions

biosData.loc[biosData["height_cm"]>175, ['name','born_date']]
````

### Key Notes:
- **First example (`athlete_df.loc[boolean_array][['name', 'born_date']]`)**: The filtering happens first, and then you select the columns from the filtered result.
- **Second example (`athlete_df.loc[boolean_array]['name', 'born_date']`)**: This fails because it’s interpreting `['name', 'born_date']` as a tuple, not a list, which causes an error.
- **Best practice**: Use `loc[boolean_array, ['column1', 'column2']]` to filter rows and select specific columns in a single operation.

### Efficiency Difference:

- **`athlete_df.loc[athlete_df["height_cm"] > 175, ['name','born_date']]`** performs both row filtering and column selection in one step.
- **`athlete_df.loc[athlete_df["height_cm"] > 175][['name','born_date']]`** filters the rows first and then selects the columns, which involves an extra step.
- **Both are correct**, but the first approach is more efficient because it filters and selects in a single step.

## **add columns** using data in DF
- Wrong way
```python
df['first_name'] = df['name'].split(' ').[0]
```
This tries to apply split(' ') directly to the entire column, not to individual elements. The split() method is a standard Python string method that operates on individual strings, not on pandas Series or DataFrames. Thus, it raises an error because it's trying to split the entire column at once, which isn't possible.
- Correct way

````python
bios_new['first_name'] = bios_new['name'].str.split(' ').str[0]
bios_new
````

## **Chaining Methods** and its Impact
Pandas allows method chaining, which can lead to more readable code but can impact memory efficiency and performance. Avoid using inplace=True with chains as it will break the chain.

````python
df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})

display(df)

# Chain methods to filter, sort, and reset index
df_filtered = df[df['A'] > 1].sort_values(by='B',ascending=False).reset_index(drop=True)

print(df_filtered)
````

### Risks of Not Using .copy()

- Unintended Side Effects: If you modify usa, you might unintentionally modify FullBios because both variables will reference the same underlying data.
- SettingWithCopyWarning: If you perform chained assignments on the usa DataFrame without using .copy(), pandas might raise a SettingWithCopyWarning. This warning occurs because pandas can't guarantee whether you're modifying a copy or a view of the original DataFrame, which can lead to ambiguous and unpredictable results

## **Memory** management
Changing data types from int64 to int32 or float64 to float32 can save significant memory without losing much precision.

````python
df = pd.DataFrame({'A': [1, 2, 3], 'B': [4.0, 5.0, 6.0]})
print(df.dtypes)
# Downcasting integers and floats to save memory
df['A'] = pd.to_numeric(df['A'], downcast='integer')
df['B'] = pd.to_numeric(df['B'], downcast='float')

print(df.dtypes)
df
````
