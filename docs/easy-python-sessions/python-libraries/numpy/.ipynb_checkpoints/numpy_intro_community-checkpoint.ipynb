## 1. What is NumPy?   
**NumPy (Numerical Python)** is the core library for numerical computing in Python.

It provides:
- Fast **n-dimensional arrays**
- Mathematical operations on arrays
- Tools for linear algebra, statistics, and random numbers

### Why NumPy over Python lists?
| Feature | Python List | NumPy Array |
|------|-----------|-------------|
| Speed | Slow | Very fast (C-backed) |
| Memory | High | Low |
| Math ops | Loop needed | Vectorized |
| Shape | Flexible | Fixed, structured |


### Identity matrix

- This line creates the 3x3 identity matrix by manually defining a nested list of values and using the general np.array() function to convert it into a NumPy array. This achieves the same result as np.eye(3) but without using the specific function designed for identity matrices.

<div align="center">
  <img src="data/9.png" alt="Alt text" width="200"/>
</div>

````python
# "Same identity matrix but without inbuilt function:"
arr = np.array([[1,0,0],[0,1,0],[0,0,1]])
print(arr)
print("Dimension:", arr.ndim)
````



- This line of code generates the 3x3 identity matrix but does not assign it to a variable or explicitly print it. In an interactive Python session (like a Jupyter notebook or a basic interpreter), the interpreter automatically displays the result of the unassigned expression.

````python
np.eye(3)
````

- This line of code generates the 3x3 identity matrix and assigns it to a variable named c. The array is stored in memory but is not displayed until a print(c) or other explicit print statement is executed.

````python
c = np.eye(3)
print(c)
print("Dimensions:", c.ndim)
````

### arange

- np.arange(start, stop, step): Returns an array with values from start to stop-step, with the specified step.
- Use np.arange() when you know the desired step size and are primarily working with integers. It behaves similarly to Python's built-in range() function.


````python
np.arange(0, 10, 2)
````

### linspace

- np.linspace(start, stop, num): Returns an array of num evenly spaced values from start to stop (inclusive by default).
- Use np.linspace() when you know the required number of elements and need the endpoint to be included, or when working with floating-point numbers to avoid potential precision issues. 

````python
np.linspace(0, 1, 5)
````

## 8. Basic Math Functions

<div align="center">
  <img src="data/5.png" alt="Alt text" width="700"/>
</div>

````python
a = np.array([1, 2, 3])
print("Shape:", a.shape)
b = np.sum(a) # ----> Show data type too
print("Sum of all elements:", np.sum(a))
print("Dimension:", b.ndim)
print("Shape:", b.shape)
````

````python
a = np.array([1,2,3])
print("Shape:", a.shape)
res3 = np.sum(a, keepdims=True)
print("Sum of all elements:",res3)
print("Dimension:",res3.ndim)
print("Shape:",res3.shape)
````


### 🔹 Comparison Table (Example with `a = np.array([1,2,3])`)

| Expression                  | Value | Type          | Shape             | ndim |
|-----------------------------|-------|---------------|-------------------|------|
| `np.sum(a)`                 | 6     | numpy.int64   | Scalar (no shape) | N/A  |
| `np.sum(a, keepdims=True)`  | [6]   | numpy.ndarray | (1,)              | 1    |
| `np.argmax(a)`              | 6     | numpy.int64   | Scalar (no shape) | N/A  |
| `np.argmax(a, axis=0)`      | [6]   | numpy.ndarray | (1,3)             | 1    |
| `np.argmin`                 | 6     | numpy.int64   | Scalar (no shape) | N/A  |
| `np.argmin(a, axis=0)`      | [6]   | numpy.ndarray | (1,3)             | 1    |
- 

---

### 🔹 Comparison Table (Example with `a = np.array([[1,2,3],[4,5,6]])`)

| Approach                | Output Example        | Shape / ndim | Best Practice Use Case |
|--------------------------|-----------------------|--------------|------------------------|
| **Scalar (default)**     | `np.sum(a)` → `21`   | Scalar (no shape, no ndim) | When you only need a single number (e.g., reporting totals, overall mean). |
| **Axis reduction**       | `np.sum(a, axis=0)` → `[5 7 9]` | `(3,)`, ndim=1 | When analyzing structure (row sums, column means, feature-wise stats). |
| **Keepdims=True**        | `np.sum(a, axis=0, keepdims=True)` → `[[5 7 9]]` | `(1,3)`, ndim=2 | When you need results aligned for broadcasting (normalization, ML pipelines). |
| **Keepdims=True**        | `np.sum(a, keepdims=True)` → `[[21]]` | `(1,1)`, ndim=2 | When you need results aligned for broadcasting (normalization, ML pipelines). |

---

### 🔹Common Aggregate Functions

| Function      | Default (no axis) | With `axis` | With `keepdims=True` |
|---------------|-------------------|-------------|-----------------------|
| `np.sum`      | Scalar            | Reduced array | Preserves dimension |
| `np.mean`     | Scalar            | Reduced array | Preserves dimension |
| `np.min` / `np.max` | Scalar      | Reduced array | Preserves dimension |
| `np.prod`     | Scalar            | Reduced array | Preserves dimension |
| `np.std` / `np.var` | Scalar      | Reduced array | Preserves dimension |


````python
a = np.array([[1,2,3],[4,5,6]])
res1=np.sum(a)
print(res1)
#print(res1.ndim)
#print(res1.shape)
````

````python
a = np.array([[1,2,3],[4,5,6]])
res2 = np.sum(a, axis=0)
print(res2)
print(res2.ndim)
print(res2.shape)
````

````python
a = np.array([[1,2,3],[4,5,6]])
res= np.sum(a, axis=0, keepdims=True)
print(res)
print(res.ndim)
print(res.shape)
````
